```python
from pulp import *
```

```python
prob = LpProblem("The Whiskas Problem", LpMinimize)
```

```python
x1 = prob.add_variable("ChickenPercent", 0, None, LpInteger)
x2 = prob.add_variable("BeefPercent", 0, None, LpContinuous)
```

```python
x1, x2
```

#### Formulate the Objective Function

The objective function becomes:

$$
\min \; 0.013x_1 + 0.008x_2
$$

#### The Constraints

The constraints on the variables are that they must sum to 100 and that the nutritional requirements are met:

$$
x_1 + x_2 = 100
$$

$$
0.100x_1 + 0.200x_2 \geq 8.00
$$

$$
0.080x_1 + 0.100x_2 \geq 6.00
$$

$$
0.001x_1 + 0.005x_2 \leq 2.00
$$

$$
0.002x_1 + 0.005x_2 \leq 0.40
$$

```python
# Objective Function
prob += 0.013 * x1 + 0.008 * x2, "Total cost of ingredients per can"
```

```python
# Constraints
prob += x1 + x2 == 100
prob += 0.100 * x1 + 0.200 * x2 >= 8.00
prob += 0.080 * x1 + 0.100 * x2 >= 6.00
prob += 0.001 * x1 + 0.005 * x2 <= 2.00
prob += 0.002 * x1 + 0.005 * x2 <= 0.40
```

```python
prob.writeLP("WhiskasModel.lp")
```

```python
prob.solve()
```

```python
print("Status:", LpStatus[prob.status])
```

```python
for v in prob.variables():
    print(v.name, "=", v.varValue)
```

```python
print("The total cost of Ingredients per can = ", value(prob.objective))
```

```python
# Creates a list of the Ingredients
Ingredients = ["CHICKEN", "BEEF", "MUTTON", "RICE", "WHEAT", "GEL"]

# A dictionary of the costs of each of the Ingredients is created
costs = {
    "CHICKEN": 0.013,
    "BEEF": 0.008,
    "MUTTON": 0.010,
    "RICE": 0.002,
    "WHEAT": 0.005,
    "GEL": 0.001,
}

# A dictionary of the protein percent in each of the Ingredients is created
proteinPercent = {
    "CHICKEN": 0.100,
    "BEEF": 0.200,
    "MUTTON": 0.150,
    "RICE": 0.000,
    "WHEAT": 0.040,
    "GEL": 0.000,
}

# A dictionary of the fat percent in each of the Ingredients is created
fatPercent = {
    "CHICKEN": 0.080,
    "BEEF": 0.100,
    "MUTTON": 0.110,
    "RICE": 0.010,
    "WHEAT": 0.010,
    "GEL": 0.000,
}

# A dictionary of the fibre percent in each of the Ingredients is created
fibrePercent = {
    "CHICKEN": 0.001,
    "BEEF": 0.005,
    "MUTTON": 0.003,
    "RICE": 0.100,
    "WHEAT": 0.150,
    "GEL": 0.000,
}

# A dictionary of the salt percent in each of the Ingredients is created
saltPercent = {
    "CHICKEN": 0.002,
    "BEEF": 0.005,
    "MUTTON": 0.007,
    "RICE": 0.002,
    "WHEAT": 0.008,
    "GEL": 0.000,
}
```

```python
prob_full = LpProblem("The Whiskas Problem", LpMinimize)
```

```python
ingredient_vars = prob.add_variable_dict("Ingr", (Ingredients,), 0, None, LpContinuous)
```

```python
prob_full += (lpSum([costs[i] * ingredient_vars[i] for i in Ingredients]),
              "Total cost of ingredients per can")
```

```python
# The five constraints are added to 'prob'
prob_full += lpSum([ingredient_vars[i] for i in Ingredients]) == 100, "PercentagesSum"
prob_full += (
    lpSum([proteinPercent[i] * ingredient_vars[i] for i in Ingredients]) >= 8.0,
    "ProteinRequirement",
)
prob_full += (
    lpSum([fatPercent[i] * ingredient_vars[i] for i in Ingredients]) >= 6.0,
    "FatRequirement",
)
prob_full += (
    lpSum([fibrePercent[i] * ingredient_vars[i] for i in Ingredients]) <= 2.0,
    "FibreRequirement",
)
prob_full += (
    lpSum([saltPercent[i] * ingredient_vars[i] for i in Ingredients]) <= 0.4,
    "SaltRequirement",
)
```

```python
prob_full.writeLP("WhiskasModel_full.lp")
```

```python
prob_full.solve()
```

```python
print("Status:", LpStatus[prob_full.status])
```

```python
for v in prob_full.variables():
    print(v.name, "=", v.varValue)
```

```python
print("Total Cost of Ingredients per can = ", value(prob_full.objective))
```
